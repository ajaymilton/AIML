## Understanding Exceptions

Exception handling in python allows you handle errors gracefully and take corrective actions without stopping the execution of the program. This lesson will cover the basics of exceptions, including how to use try, except, else, and finally blocks.

### What are Exceptions?

Exceptions are events the disrupt the normal flow of a program. They occur when an error is encountered during program execution. Common exceptions include:

    => ZeroDivisionError: Dividing by  zero.
    => FileNotFoundError: File not found.
    => ValueError: Invalid value.
    => TypeError: Invalid type.

```python
## Exception try, except block

try:
    a=b
except:
    print("The variable has not been assigned.")
```

```python
try:
    a=b
except NameError as ex: ## NameError is a class of exception. ex is an object of NameError class
    print(ex)
```

```python
result = 10/0
```

```python
try:
    result = 10/0
except ZeroDivisionError as ex:
    print(ex)
    print("You cannot divide a number by zero.") ## multiple statements can be executed in except block
```

```python
try: 
    result = 1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("You cannot divide a number by zero.")
```

```python
## Exception Main block, the parent class of all exceptions is Exception. It will catch all exceptions.

try: 
    result = 1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
except Exception as ex1: ## Exception is a base class of all exceptions. It will catch all exceptions
    print(ex1)
    print("Main exception block executed.") ## multiple statements can be executed in except block
```

```python
try:
    num = int(input("Enter a number: "))
    result = 10/num
except ValueError:
    print("Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("You cannot divide a number by zero.")
except Exception as ex:
    print(ex)
```

```python
## try, except, else block

try:
    num = int(input("Enter a number: "))
    result = 10/num
except ValueError:
    print("Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("You cannot divide a number by zero.")
except Exception as ex:
    print(ex)
else:
    print(f"The result is: {result}")
```

```python
## try, except, else and finally block

try:
    num = int(input("Enter a number: "))
    result = 10/num
except ValueError:
    print("Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("You cannot divide a number by zero.")
except Exception as ex:
    print(ex)
else:
    print(f"The result is: {result}")
finally:
    print("Execution completed.")
```
