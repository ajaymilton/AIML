## Sets
Sets are unordered collections of unique elements. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. They are useful for membership testing, removing duplicates from a sequence, and performing mathematical operations like union, intersection, and difference.

```python
## create a set

my_set = {1, 2, 3, 4, 5}
print(my_set)  # Output: {1, 2, 3, 4, 5}
print(type(my_set))  # Output: <class 'set'>
```

```python
my_empty_set = set()  # Creating an empty set
print(my_empty_set)  # Output: set()
```

```python
my_set = set([1, 2, 3, 4, 5])  # Creating a set from a list
print(my_set)  
```

```python
my_empty_set = set([1, 2, 3, 4, 2, 4, 5])  # Creating a set from a list with duplicates
print(my_empty_set)
```

```python
## basic set operations
## Adding and removing elements

my_set.add(6)  # Adding an element to the set
print(my_set)  # Output: {1, 2, 3, 4
```

```python
## Removing an element from the set

my_set.remove(3)  # Removing an element from the set
print(my_set)  # Output: {1, 2, 4, 5
```

```python
my_set.remove(10)  # Attempting to remove an element that doesn't exist will raise a KeyError
```

```python
my_set.discard(10)  # Discarding an element that doesn't exist will not raise an error
print(my_set)  # Output: {1, 2, 4, 5}
```

```python
## pop method removes and returns an arbitrary element from the set. If the set is empty, it raises a KeyError.

removed_element = my_set.pop()  # Removing an arbitrary element from the set
print(f"Removed element: {removed_element}")  # Output: Removed element: <some
print(my_set)  # Output: {1, 2, 4, 5} (the set after removing an element)   
```

```python
## clearing a set

my_empty_set.clear()  # Clearing all elements from the set
print(my_empty_set)  # Output: set() (the set is now empty)
```

```python
# set membership test

my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  # Output: True
print(6 in my_set)  # Output: False
```

```python
## Mathematical set operations

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

### union
union_set = set1.union(set2)  # Union of two sets
print(union_set)  # Output: {1, 2, 3, 4, 5, 6, 7, 8}

### intersection
intersection_set = set1.intersection(set2)  # Intersection of two sets
print(intersection_set)  # Output: {4, 5}

set1.intersection_update(set2)  # Update set1 with the intersection of set1 and set2
print(set1)  # Output: {4, 5}
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

## difference
print(set1.difference(set2))  # Output: {1, 2, 3}
```

```python
## symmetric difference

set1.symmetric_difference(set2)  # Output: {1, 2, 3, 6, 7, 8}
```

```python
## sets methods
set1 = {1, 2, 3, 4, 5 ,6}  
set2 = {4, 5, 6,}

## issubset
print(set1.issubset(set2))  # Output: False
print(set2.issubset(set1))  # Output: True

print(set1.issuperset(set2))  # Output: True
print(set2.issuperset(set1))  # Output: False
```

```python
list = [1, 2, 2, 4, 3, 4, 5]
set_from_list = set(list)  # Creating a set from a list with duplicates
print(set_from_list)  # Output: {1, 2, 3, 4, 5} (duplicates are removed)
```

```python
### Counting unique words in a text using a set

text = "This is a sample text with some repeated words. This text is just a sample."
words = text.split()  # Splitting the text into words
unique_words = set(words)  # Creating a set of unique words
print(unique_words)  # Output: {'This', 'is', 'a', 'sample', 'text', 'with', 'some', 'repeated', 'words.', 'just'} (unique words)
print(len(unique_words))  # Output: 8 (number of unique words)
```
