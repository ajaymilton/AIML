# Introduction to Tuples

Tuples are similar to lists in Python, but they are immutable, meaning that once a tuple is created, its elements cannot be changed, added, or removed. Tuples are defined using parentheses `()` instead of square brackets `[]`.

```python
## creating a tuple 

my_tuple = ()  
print(my_tuple)
print(type(my_tuple))
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
## type conversion

numbers = tuple([1, 2, 3, 4, 5])
numbers
```

```python
list((1, 2, 3, 4, 5))
```

```python
mixed_tuple = (1, 2.5, "Hello", True)
print(mixed_tuple)
```

```python
## Accessing elements in a tuple
numbers = (1, 2, 3, 4, 5)
```

```python
numbers[0]  # Accessing the first element
```

```python
print(numbers[-1])  # Accessing the last element
print(numbers[1:4])  # Accessing a slice of the tuple
print(numbers[:3])  # Accessing the first three elements
print(numbers[2:])  # Accessing elements from the third element to the end
```

```python
## tuple operations

numbers[::-1]  # Reversing the tuple
```

```python
concatenated_tuple = numbers + mixed_tuple  # Concatenating two tuples
print(concatenated_tuple)
```

```python
mixed_tuple * 3
```

```python
numbers * 3
```

```python
## immmutable nature of tuples
## tuples are immutable, meaning their elements cannot be changed after creation. However, you can create a new tuple by concatenating or slicing existing tuples.

lst=[1, 2, 3]
print(lst)

lst[0] = 'Ajay'
print(lst)  
```

```python
numbers[1] = 10  # This will raise a TypeError because tuples are immutable
```

```python
## tuple methods

print(numbers.count(2))  # Count the occurrences of 2 in the tuple
print(numbers.index(3))  # Find the index of the first occurrence of 3 in the tuple
```

```python
## packing and unpacking tuples
## packing
packed_tuple = 1, 2, 3, 4, 5  # Packing values into a tuple
print(packed_tuple)
```

```python
## unpacking

a, b, c, d, e = packed_tuple  # Unpacking the tuple into variables
print(a, b, c, d, e)
```

```python
## unpacking with *

numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers  # Unpacking with *
print(first)  # Output: 1
print(middle)  # Output: [2, 3, 4]
print(last)  # Output: 5
```

```python
## nested list

lst = [[1, 2 ,3 ,"hello",3.24 ,4], [3, 4], [5, 6]]
print(lst[0][1])  # Accessing the second element of the first list
print(lst[1][0])  # Accessing the first element of the second list
print(lst[0][0::2])  # Accessing a slice of the third list
```

```python
## nested tuples
nested_tuple = ((1, 2, 3), (4, "a", 6), (True, False))

print(nested_tuple[0][1])  # Accessing the second element of the first tuple
print(nested_tuple[1][1])  # Accessing the second element of the second tuple
print(nested_tuple[2][0])  # Accessing the first element of the third tuple
```

```python
## iterating through a tuple

for sub_tuple in nested_tuple:
    for element in sub_tuple:
        print(element,end=" ")
    print()  # Print a newline after each sub-tuple
```

### Conclusion
Tuples are a fundamental data structure in Python that allow you to store ordered collections of items.
