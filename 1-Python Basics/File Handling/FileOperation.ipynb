## File Operation - Read and Write Files

File handling is a crucial part of any programming language. Python provides built-in functions and methods to read and write to files, both text and binary. This lesson will cover the basics of file handling, including reading and writing text files  and binary files

```python
### Read a whole file

with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
### Read a file line by line

with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())  # Use strip() to remove leading/trailing whitespace or removes new line character
```

```python
### writing a file

with open("example.txt", "w") as file:
    file.write("this a new line\n")
    file.write("this is another new line\n")
```

```python
## write a file without overwriting the existing content

with open("example.txt", "a") as file:
    file.write("this is a new line append operation\n")
    file.write("this is another appended new line\n")
```

```python
## writing alist of lines to a file

lines = ["This is line 1\n", "This is line 2\n", "This is line 3\n"]
with open("example.txt", "a") as file:
    file.writelines(lines)
```

```python
### Binary files
## Writing binary data to a file

data = b'\x00\x01\x02\x03\x04\x05'
with open("example.bin", "wb") as file:
    file.write(data)
```

```python
##Reading binary data from a file   

with open("example.bin", "rb") as file:
    content  = file.read()
    print(content)
```

```python
### read the content from a source text file and write it to a destination file
# Copy a text file
with open("example.txt", "r") as source_file:
    content = source_file.read()

with open("destination.txt", "w") as destination_file:
    destination_file.write(content)
```

```python
## Assignment 
### read a text file and count the no. of lines, words, and characters
# counting lines, words, and characters in a text file

def count_text_file(file_path):
    with open(file_path, "r") as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = "example.txt"
lines, words, characters = count_text_file(file_path)
print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```

The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exists, its content is truncated (ie, the file is overwritten)

```python
### writing and the reading a file 

with open("example.txt", "w+") as file:
    file.write("This is a new line\n")
    file.write("This is another new line\n")
    
    file.seek(0)  # Move the cursor to the beginning of the file
    content = file.read()
    print(content)
```
