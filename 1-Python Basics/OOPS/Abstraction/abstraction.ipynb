### Abstraction
Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This helps in reducing programming complexityand effort.

- Hides implementation details
- Exposes essential functionality
- Uses abstract classes and methods
- Focuses on what functionality is exposed

In simple words:
Abstraction focuses on what an object does rather than how it does it.

Python provides the abc module (Abstract Base Classes) to implement abstraction.

Important concepts:
- Abstract Class
- Abstract Method
- Concrete Class

#### Abstract Class
An abstract class is a class that cannot be instantiated directly.It acts as a blueprint for other classes.
#### Abstract Method
An abstract method is a method declared in an abstract class but without a complete implementation.The child class must implement it.

- The abstract class defines what functionality every method must provide.
- The child classes define how that functionality works.

```python
from abc import ABC, abstractmethod

## Abstract base class
class Vehicle(ABC):
    def drive(self):
        print("the vehicle is driving")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car engine started.")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car = Car()
operate_vehicle(car)
```

```python
## Abstraction example

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass   

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

circle = Circle(5)
rectangle = Rectangle(4, 6) 

print(f"Area of Circle: {circle.area()}")
print(f"Area of Rectangle: {rectangle.area()}")
```
