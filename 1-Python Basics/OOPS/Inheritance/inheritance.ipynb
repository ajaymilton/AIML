## Inheritance in Python
Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance, demonstrating how to create and use them in Python.

#### What is inheritance?
Inheritance is a concept in which one class can acquire the attributes and methods of another class.

In simple words, a child class can reuse the code written inside a parent class without writing it again.

#### What is super() in inheritance?
super() is used to access methods from the parent class.

#### What if the child class has its own attributes?
A child class can have everything from its parent and also define its own attributes and methods.

#### What is method overriding?
Method overriding means a child class provides its own implementation of a method that already exists in the parent class.

### Types of inheritance in Python
Python supports five commonly discussed inheritance types.
1. Single inheritance => One child inherits from one parent.
2. Multiple inheritance => One child inherits from multiple parent classes.
3. Multilevel inheritance => A class inherits from a child class, creating a chain of inheritance.
4. Hierarchical inheritance => Multiple child classes inherit from the same parent class.
5. Hybrid inheritance => Hybrid inheritance combines two or more inheritance types.

```python
## Inheritance (Single Inheritance)
## Parent class 

class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car.")
```

```python
car1 = Car(4,4,"petrol")
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,selfdriving):
        super().__init__(windows,doors,enginetype)
        self.is_selfdriving = selfdriving

    def selfdriving(self):
        print(f"Tesla supports self driving feature: {self.is_selfdriving}")
```

```python
tesla1 = Tesla(4,4,"electric",True)
tesla1.selfdriving()
tesla1.drive()
```

```python
## Multiple Inheritance
## When a class inherits from more than one base class, it is called multiple inheritance. The derived class inherits the features of all the base classes.

## base class 1
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Subclass must implement this method.")

## base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner

## Derived class
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} says Woof!"

# create a object

dog = Dog("Buddy", "Aldo")
print(dog.speak())
print(f"Owner: {dog.owner}")
    
```
