### Generators
Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

### What is a Generator?
A Generator is a special type of iterator that generates values one at a time instead of storing all values in memory.

In Python, we create generators using:
1. Generator functions (yield).
2. Generator expressions.
Let's first understand the difference.
#### Normal function vs Generator function
Normal function:

def get_numbers():    
    return [1, 2, 3, 4, 5]

result = get_numbers()

print(result)

Output:
[1, 2, 3, 4, 5]

Here, the function creates and returns the entire list at once.

Generator function:

def get_numbers():   
    yield 1   
    yield 2    
    yield 3    
    yield 4    
    yield 5

result = get_numbers()

print(result)

Output:
<generator object get_numbers at ...>

Why?

Because yield doesn't return all values immediately. It produces values one at a time when requested.

### Understanding yield vs return
This is the most important concept in Generators.
#### return	
- Returns a value and exits the function	
- Function execution ends	
- Usually used in normal functions	
- Doesn't preserve execution state after returning	
#### yield
- Produces a value and pauses the function
- Function execution can resume
- Used in generator functions
- Preserves local state between yields

`The key point:` A generator remembers its previous execution state, including local variables and where it paused.

`Remember:` Every generator is an iterator, but not every iterator is a generator.

#### Iterator
- An object implementing the iterator protocol
- Created using iter() or a custom class  
- Requires __iter__() and __next__()
- State management may need manual coding
- Memory usage depends on implementation

#### Generator
- A special type of iterator
- Created using yield or generator expressions'
- Python implements these automatically
- State is maintained automatically
- Often memory efficient

```python
def square(n):
    for i in range(3):
        yield i ** 2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)

a
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen = my_generator()

gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

### Practical Example: Reading Large Files
Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.

```python
### Practical : Reading large files using generators

def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
