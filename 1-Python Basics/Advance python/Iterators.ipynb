### Iterators
Iterators are advanced python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

- An Iterator in Python is an object that allows us to access elements one at a time from a collection, such as a list, tuple, string or dictionary.

- Instead of accessing all elements at once, an iterator gives us the next element whenever we request it.

- Iterators are particularly useful when working with large datasets in Data Science and AIML.

Iterable: An object that can be looped over.
Examples:
- List
- Tuple
- String
- Dictionary
- Set

Iterator: An object that produces elements one by one when requested.

#### Understanding iter() and next()
Python provides two important built-in functions.

iter() => Converts an iterable into an iterator.

next() => Returns the next element from an iterator.
#### StopIteration exception
Python raises a StopIteration exception when there are no more elements. This is how Python knows that an iterator has finished.

### Why are Iterators important in AIML?
- Imagine you have a dataset containing 10 million records.
- Loading and processing everything at once can consume significant memory.
- Iterators allow you to process records one at a time.

```python
my_list = [1, 2, 3, 4, 5, 6]
for i in my_list:
    print(i)
```

```python
# String Iterators

my_string = "Hello, World!"
string_iterator = iter(my_string)

print(next(string_iterator))  # Output: 'H'
print(next(string_iterator))  # Output: 'e'
```
