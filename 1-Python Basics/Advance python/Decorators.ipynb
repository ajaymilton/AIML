### Decorators
Decorators are a powerful and flexible feature in Python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

### Function Copy in Python
In Python, functions are first-class objects.

This means functions can be:
1. Assigned to variables.
2. Passed as arguments to other functions.
3. Returned from other functions.
4. Stored inside lists and dictionaries.

#### Assigning a function to another variable
Consider this example:

def greet():

print("Hello, Welcome to Python!")

#Assigning function to another variable

say_hello = greet

greet()

say_hello()

Output:

Hello, Welcome to Python!
Hello, Welcome to Python!


What's happening here?
- greet is a function.
- say_hello = greet assigns the same function object to another variable.
- Both variables refer to the same function.

#### Passing a function as an argument
Python allows you to pass one function into another function.

def greet():    

return "Hello!"

def display(func):    

message = func()    

print(message)

display(greet)

Output:

Hello!

Explanation:
- greet is passed as an argument.
- display() receives it through the parameter func.
- func() executes the original function.

Notice that we pass greet, not greet().

#### Returning a function from another function
A function can also return another function.

def outer_function():    

def inner_function():        

print("This is an inner function")    

return inner_function

result = outer_function()

result()

Output:

This is an inner function

This concept is particularly important because it leads us to Closures and Decorators.

#### Python has copy.copy() and copy.deepcopy().
Python functions are not duplicated by these operations. They return the same function object.

- new_func = func => Another reference to the same function
- copy.copy(func) => Same function object
- copy.deepcopy(func) => Same function object

### Closure
A Closure is a function that remembers and retains access to variables from its enclosing (outer) function, even after the outer function has finished executing.

```python
def outer_function():

    message = "Hello Python"

    def inner_function():
        print(message)

    return inner_function


result = outer_function()

result()
```

#### Understanding how Closures work
Let's break down the execution.
Closure execution flow
1. outer_function() executes
   
   A variable message = "Hello Python" is created.
2. inner_function() is created
   
   It uses the variable message from the outer function.
3. inner_function is returned
   
   The outer function finishes execution, but the inner function retains access to message.
4. result() executes
   
   The inner function accesses the remembered variable and prints Hello Python.

   Closure:
   - Deals with enclosing variables
   - Inner functions retain outer scope
   - Supports state retention
   - Example: nested functions

```python
### Decorator

def main_welcome(func):
    
    def sub_welcome():
        print("Welcome to Advance Python")
        func()
        print("please enjoy the course")
    return sub_welcome()
```

```python
def course_intro():
    print("This is a course on Advance Python")

course_intro()
```

```python
main_welcome(course_intro)
```

```python
@main_welcome
def course_intro():
    print("This is a course on Advance Python")

```

```python
## Decorator

def my_decorator(func):
    def wrapper():
        print("Something is happening before the funtion is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
## decorator with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Hello")
```

```python
say_hello()
```
